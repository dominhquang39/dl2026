```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('data/lr.csv', header=None)
X = df.iloc[:, [0]].values
y = df.iloc[:, [1]].values

print(f'Shape of X: {X.shape}')
print(f'Shape of y: {y.shape}')
```

```python
def loss_function(N, X_bar, w, y):
    y_hat = X_bar.dot(w) 
    loss = (1/(2*N)) * np.sum((y_hat - y)**2)
    return loss

def gradient(N, X_bar, w, y):
    y_hat = X_bar.dot(w)
    gradient = (1/N) * X_bar.T.dot(y_hat - y)
    return gradient

def gradient_descent(X, y, lr, threshold):
    N = len(y)
    X_bar = np.concatenate((np.ones((N, 1)), X), axis=1) # (N, 2)
    w = np.array([[0.0], [1.0]]) # w0 = 0.0 and w1 = 1.0

    count = 0
    converged = False

    while not converged:
        w_old = w.copy()
        grad = gradient(N, X_bar, w, y)
        w = w_old - lr * grad
        diff = np.sqrt(np.sum((w - w_old)**2))
        count += 1
        if count % 100 == 0:
            loss = loss_function(N, X_bar, w, y)
            print(f"Iteration {count}: Loss={loss:.3f}, Difference between old and new weights: {diff}")

        if diff < threshold:
            converged = True
            print(f"Model converge after {count} iterations")
            
    print(f"Final weights:")
    print(f"w0 = {w[0][0]:.3f}, w1 = {w[1][0]:.3f}")
    return w
```

```python
learning_rate = 0.0005
threshold = 1e-6
w = gradient_descent(X, y, learning_rate, threshold)
```

```python
def predict(X, w):
    return w[0][0] + w[1][0] * X

new_X = np.array([[80], [165], [170]])
predicted_y = predict(new_X, w)
print(f"Predicted y for new Xs: {predicted_y.flatten()}")
```
