```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('house.csv')
X = df[['Area']].values
y = df[['Price']].values

print(f'Shape of X: {X.shape}')
print(f'Shape of y: {y.shape}')
```

```python
def loss_function(N, X_bar, y, w):
    y_hat = X_bar.dot(w)
    loss = (1 / (2*N)) * np.sum((y_hat - y)**2)
    return loss

def gradient(N, X_bar, y, w):
    y_hat = X_bar.dot(w)
    gradient = (1/N) * X_bar.T.dot(y_hat - y)
    return gradient

def gradient_descent(X, y, lr, threshold=1e-6):
    N = len(y)
    X_bar = np.concatenate((np.ones((N, 1)), X), axis=1)
    w = np.array([[0.0], [1.0]])

    count = 0
    converged = False

    while not converged:
        w_old = w.copy()
        grad = gradient(N, X_bar, y, w)
        w = w_old - lr * grad
        diff = np.sqrt(np.sum(w - w_old)**2)
        count += 1
        if count % 100 == 0:
            loss = loss_function(N, X_bar, y, w)
            print(f"Iteration {count}: Loss={loss:.3f}, Difference between old and new weights: {diff}")

        if diff <= threshold:
            converged = True
            print(f"Model converged after {count} iterations")
    
    print('Final weights:')
    print(f'w0 = {w[0][0]:.3f}, w1 = {w[1][0]:.3f}')
    return w
```

```python
learning_rate = 0.001
threshold = 1e-6
w = gradient_descent(X, y, lr=learning_rate, threshold=threshold)
```

```python
def predict(X, w):
    N = len(X)
    X_bar = np.concatenate((np.ones((N, 1)), X), axis=1)
    y_hat = X_bar.dot(w)
    return y_hat

new_X = np.array([[50], [80], [120]])
y_pred = predict(new_X, w)
print('Predicted prices:', y_pred.flatten())
```
